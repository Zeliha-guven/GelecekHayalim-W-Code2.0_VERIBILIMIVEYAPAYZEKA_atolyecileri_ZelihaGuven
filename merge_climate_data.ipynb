```python
"""
Import core libraries and set display options for climate data analysis.
Includes data handling, visualization, and extended table view for comprehensive inspection.
"""
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure display options for better visibility
pd.set_option('display.max_columns', 500)
pd.set_option('display.max_rows', 500)
```

## Load climate-related datasets

```python
df = pd.read_csv("data/final_iklim_analizi.csv")
git = pd.read_csv("data/owid-energy-data.csv")
co = pd.read_csv("data/owid-co2-data.csv")
```

## Indicator Analysis and Feature Engineering

```python
co["person_gdp"] = co["gdp"] / co["population"]  # Calculate GDP per capita
```

```python
# Preview the CO₂ dataset
co.head()
```

```python
# Select relevant columns for CO₂ and energy analysis
columns_to_keep = [
    'country',           # Country name
    'year',              # Observation year
    'iso_code',          # Country ISO code
    'co2_per_capita',    # CO₂ emissions per person
    'cumulative_co2',    # Total historical CO₂ emissions
    'energy_per_capita', # Energy use per person
    'share_global_co2',  # Country’s share in global CO₂
    'person_gdp',        # GDP per capita
    'population'         # Total population
]
```

```python
# Keep only relevant columns for climate and socioeconomic analysis
co = co[columns_to_keep]
```

```python
# Drop columns deemed non-essential for the final readiness model
df.drop(["social", "water", "food"], inplace=True, axis=1)
```

```python
# Filter Git dataset to keep key energy-related indicators
git = git[["iso_code", "year", "other_renewables_share_elec", "country"]]
```

```python
# Filter data to match the desired year range for consistency across datasets
git = git[(git["year"] >= 1995) & (git["year"] < 2024)]
```

```python
# Apply the same year filter to the CO₂ dataset for consistency
co = co[(co["year"] >= 1995) & (co["year"] < 2024)]
```

```python
co.head()
```

```python
# Rename 'country' column to 'iso_code' for consistency in primary key usage
df["iso_code"] = df["country"]
```

```python
# Drop the original 'country' column after standardizing ISO codes
df.drop("country", inplace=True, axis=1)
```

```python
df.head()
```

```python
# Merge all datasets on 'iso_code' and 'year' to build the integrated climate model base
merged_df = df.merge(co, on=['iso_code', 'year'], how='left')
merged_df = merged_df.merge(git, on=['iso_code', 'year'], how='left')
```

```python
merged_df.shape
```

```python
merged_df.head()
```

```python
merged_df.columns
```

```python
# Drop the duplicate 'country_y' column created during merging
merged_df.drop("country_y", inplace=True, axis=1)
```

```python
merged_df.isnull().sum()
```

```python
# Drop columns with entirely missing values from the merged dataset
merged_df.drop(["co2_per_capita_x", "cumulative_co2_x"], inplace=True, axis=1)
```

```python
# Save the final cleaned and merged dataset for modeling and analysis
merged_df.to_csv('data/final.csv', index=False)
```
